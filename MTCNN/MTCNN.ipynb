```python
import cv2
from mtcnn import MTCNN
```

```python
def crop_faces(image_path, output_prefix="face"):

    detector = MTCNN()

    image = cv2.imread(image_path)
    if image is None:
        print(f"Error: Could not load image from {image_path}")
        return


    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)


    detections = detector.detect_faces(image_rgb)

    if not detections:
        print("No faces detected.")
        return

    print(f"Detected {len(detections)} face(s). Processing crops...")


    for i, face in enumerate(detections):
        # Extract bounding box coordinates
        x, y, width, height = face['box']

        # Ensure coordinates are within image boundaries (prevents negative index errors)
        x, y = max(0, x), max(0, y)

        # Crop the face using NumPy slicing
        cropped_face = image[y:y+height, x:x+width]

        output_filename = f"{output_prefix}_{i}.jpg"
        cv2.imwrite(output_filename, cropped_face)
        print(f"Saved: {output_filename}")

```

```python
crop_faces("tom-new.jpg")
```
